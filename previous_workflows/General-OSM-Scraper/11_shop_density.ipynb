```python
## Cell 1 · Imports

import pandas as pd
import numpy as np
from sklearn.neighbors import BallTree

df = pd.read_csv("csv/00_base_data.csv")
print(f"Loaded {len(df)} records")
```

```python
## Cell 2 · Count Shops Within 100m

shop_coords = np.radians(df[["lat", "lon"]].values)
tree = BallTree(shop_coords, metric="haversine")

# 100m in radians
radius = 100 / 6371000

counts = tree.query_radius(shop_coords, r=radius, count_only=True)

# Subtract 1 to exclude the shop itself
df["shops_within_100m"] = counts - 1

print(f"✓ Done")
print(f"  Mean : {df['shops_within_100m'].mean():.1f}")
print(f"  Min  : {df['shops_within_100m'].min():.0f}")
print(f"  Max  : {df['shops_within_100m'].max():.0f}")
```

```python
## Cell 3 · Save

df_out = df[["osm_id", "shops_within_100m"]]
df_out.to_csv("csv/11_shop_density.csv", index=False, encoding="utf-8")
print(f"Saved {len(df_out)} records to csv/11_shop_density.csv")
print(df_out.describe().round(1))
```
